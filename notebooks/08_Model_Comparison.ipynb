### 2. Operational Trade-Offs: Latency vs. Detection Accuracy
- **Random Forest & XGBoost:** Deliver the highest Macro-F1 and lowest False Positive Rates, making them ideal for high-confidence security alerts.
- **Decision Tree & Logistic Regression:** Offer sub-millisecond classification latencies, appropriate for inline packet inspection gateways with stringent throughput constraints.
