### 1. Visualizing Network Traffic Signatures
Exploratory Data Analysis visualizes the statistical differences between benign baseline communications and malicious cyberattacks.

```
import os
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()))

import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import load_dataset
from src.visualization import (
    plot_class_distribution_sns,
    plot_benign_vs_attack_sns,
    plot_flow_metrics_distribution_sns,
    plot_correlation_heatmap_sns,
)

df, _ = load_dataset(use_sample_if_empty=True)
print(f"Loaded {len(df):,} flows for exploratory data analysis.")
```

### 2. Overall Class Distribution
Visualizing the distribution of benign traffic and individual attack classes.

```
p1 = plot_class_distribution_sns(df)
img = plt.imread(p1)
plt.figure(figsize=(10, 6))
plt.imshow(img)
plt.axis('off')
plt.show()
```

**Analysis:** Malicious attacks exhibit a pronounced long-tail distribution. High-volume attacks such as DoS Hulk and PortScan generate substantial flow counts, while stealthy intrusion attempts (e.g. Web Attacks, Botnet C2) generate smaller flow footprints.

### 3. Benign vs Malicious Threat Traffic Proportions

```
p2 = plot_benign_vs_attack_sns(df)
img2 = plt.imread(p2)
plt.figure(figsize=(12, 5))
plt.imshow(img2)
plt.axis('off')
plt.show()
```

### 4. Heavy-Tailed Network Flow Metrics
Examining Flow Duration, Byte Rates, Packet Rates, and Packet Length distributions.

```
p3 = plot_flow_metrics_distribution_sns(df)
img3 = plt.imread(p3)
plt.figure(figsize=(14, 10))
plt.imshow(img3)
plt.axis('off')
plt.show()
```

**Analysis:** Network flow duration and byte rates span several orders of magnitude (from sub-millisecond microbursts to hours-long persistent TCP sessions), demonstrating the necessity of logarithmic transforms and quantile clipping in machine learning pipelines.

### 5. Flow Feature Correlation Matrix

```
p4 = plot_correlation_heatmap_sns(df)
img4 = plt.imread(p4)
plt.figure(figsize=(12, 10))
plt.imshow(img4)
plt.axis('off')
plt.show()
```

**Analysis:** Several feature groups exhibit near-perfect correlation (e.g., Subflow Forward Packets vs Total Forward Packets; Average Packet Size vs Mean Packet Length). Recognizing collinearity informs feature selection.
